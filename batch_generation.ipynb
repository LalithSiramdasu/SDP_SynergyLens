# Batch File Generation
Generate random batch CSV files from `datasets/train_500_rows.csv` and save them into the `batch` folder.

```python
import pandas as pd
from pathlib import Path
import numpy as np
```

```python
project_root = Path('c:/Users/HP/Desktop/SDP_SynergyLens')
batch_folder = project_root / 'batch'
batch_folder.mkdir(exist_ok=True)
batch_folder
```

```python
csv_path = project_root / 'datasets' / 'train_500_rows.csv'
df = pd.read_csv(csv_path)
df.shape
```

```python
shuffled = df.sample(frac=1, random_state=42)
shuffled = shuffled.rename(columns={
    'drug1': 'NSC1',
    'drug2': 'NSC2',
    'cell line': 'CELLNAME'
})

batch_sizes = list(range(10, 90, 10))
for size in batch_sizes:
    out_path = batch_folder / f'batch_{size}rows.csv'
    shuffled.iloc[:size].to_csv(out_path, index=False)
    print(f'Wrote {out_path} ({size} rows)')

print('Saved batch files with columns:', list(shuffled.columns))
```
